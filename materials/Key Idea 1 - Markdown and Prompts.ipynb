## Why This Module Matters

When you give a vibe coding AI a plain sentence like "build me a login system," the AI has to guess your intent. It may produce insecure code, miss requirements, or hallucinate entire APIs that do not exist.

**Markdown is not just formatting** - it is a structured signal layer that changes how LLMs process your request. Every header, bullet, and code block shifts the model's internal attention weights toward what actually matters.

```
[Plain Text Prompt]              [Structured Markdown Prompt]
"build a login system"           # Login System Feature
                                 ## Requirements
                                 - Validate username + password
                                 - Return JWT token on success
                                 ## Constraints
                                 - Language: Python
                                 - No hardcoded secrets
                                 ## Expected Output
                                 login("alice", "pass")  -> JWT string

LLM sees: ambiguous blob         LLM sees: clear task tree with sections
```

## The Prompt Economics Angle

Every token costs money at API rate. A well-structured prompt produces the correct code on the first attempt, eliminating expensive revision cycles. At enterprise scale - thousands of prompts per day - this difference in first-pass accuracy is a significant operational cost.


## Key Terms

| Term | Definition |
|---|---|
| **Token** | The smallest unit an LLM processes - roughly 3/4 of a word. Prompt length in tokens equals API cost. |
| **Attention Mechanism** | The internal weighting system that tells an LLM which parts of the prompt to focus on when generating each output token. |
| **Prompt Engineering** | The practice of crafting structured inputs to guide LLM behavior toward a desired output. |
| **Context Window** | The maximum number of tokens an LLM can process in a single request - typically 8k to 200k tokens. |
| **Markdown** | A lightweight markup language that uses symbols like #, *, and backtick fences to add semantic structure to plain text. |

## Lab 1: Simulating LLM Attention with a Prompt Parser

### Objective

Build a Python function that simulates how an LLM parser processes Markdown prompts. It detects structural signals - headers, code blocks, and bullet points - and reports what the model's attention mechanism would prioritize.


````python
import re

def simulate_llm_attention_parser(prompt_text):
    print("Simulating LLM Attention on your prompt...")

    # Headers signal high-priority task nodes
    headers = re.findall(r'^#+\s+(.*)', prompt_text, re.MULTILINE)

    # Code fences mark explicit structural boundaries
    code_blocks = re.findall(r'```python(.*?)```', prompt_text, re.DOTALL)

    print("\n--- Attention Analysis Report ---")
    if headers:
        print(f"  High Attention Nodes (Headers): {headers}")
    if code_blocks:
        print(f"  Code Boundaries Detected: {len(code_blocks)} block(s) found")
    if not headers and not code_blocks:
        print("  Warning: Low Semantic Structure - LLM may hallucinate or lose focus")

    bullet_count = len(re.findall(r'^\s*[-*]\s+', prompt_text, re.MULTILINE))
    print(f"  Bullet points (medium attention): {bullet_count}")
    print(f"  Total characters: {len(prompt_text)}")


# Test 1: A well-structured Markdown prompt
structured_prompt = """
# User Profile Lookup Feature

## Requirements
- Query database by username
- Return email and role fields
- Handle missing users gracefully

## Constraints
- Language: Python
- No hardcoded credentials
- Use parameterized queries

## Expected Output
```python
get_user_profile("alice")  # -> {"email": "alice@undp.org", "role": "Admin"}
```
"""

print("=== Test 1: Structured Prompt ===")
simulate_llm_attention_parser(structured_prompt)

# Test 2: A vague, unstructured prompt
unstructured_prompt = "build me something that looks up users in the database and returns their email and role"

print("\n=== Test 2: Unstructured Prompt ===")
simulate_llm_attention_parser(unstructured_prompt)
````

## What Just Happened?

The parser found 4 headers and 1 code block in the structured prompt - structural signals that direct the LLM's attention toward key requirements, constraints, and expected behavior.

The unstructured prompt produced zero findings. It is a single ambiguous sentence with no semantic anchors. An LLM receiving this must guess your intent, which is exactly where hallucination and security vulnerabilities are introduced.

In a real IDE like Cursor or GitHub Copilot, your prompt goes through a similar internal parsing step before the LLM generates code. Markdown structure is not cosmetic - it directly shapes output quality.
